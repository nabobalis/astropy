# WCS ticks on a bqplot image, laid out by astropy's matplotlib-free core

Each figure is a bqplot `Figure` with an image mark and no bqplot axes. The frame, ticks, grid lines and labels are bqplot marks, placed by `astropy.visualization.wcsaxes._layout` from the `wcsaxes-layout-core` branch. Drag to pan and scroll to zoom: each change of the image scales reruns the layout in the kernel.

Run this from this directory, in an environment with that branch of astropy and bqplot.

```python
from wcs_bqplot import figure
```

## IRIS-like helioprojective, rolled by 20°

The image is drawn from world coordinates: the blob sits on the crossing of the −300″ and 200″ grid lines.

```python
figure("hpc")
```

## All-sky galactic plate carrée

Longitude runs 180° → 0° → 180°, wrapping from 0° to 360° at the centre. Past the edge of the image there is no sky, and the ticks stop there.

```python
figure("car")
```

## Plain TAN

Right ascension is in hours. The formatter writes the h, m and s superscripts in mathtext, which bqplot cannot render, so the demo swaps them for Unicode letters.

```python
figure("tan")
```

## What the core returned

`fig.wcs_layout` keeps the last layout. For each coordinate it holds the placed ticks (`placed`), the tick labels after sorting and simplifying (`labels`), their anchors in display pixels (`xy`), the labels kept for drawing (`kept`), the grid lines (`grid`) and the axis label position and rotation (`axis_label`).

```python
fig = figure("hpc")
solar_x = fig.wcs_layout.coords[0]
solar_x.placed.major.world, [solar_x.labels.text[axis][i] for axis, i, _ in solar_x.kept]
```
